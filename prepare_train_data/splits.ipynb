```python
import pandas as pd
from sklearn.model_selection import train_test_split

merged_rna = pd.read_parquet('/data/merged_rna_mature.parquet') # GTEx TCGA merged mature mRNA
merged_mir = pd.read_parquet('/data/merged_mir_mature.parquet') # GTEx TCGA merged mature miRNA

# this data will be used for split and training along with scRNA data

merged_rna = merged_rna.T
merged_mir = merged_mir.T
```

```python
# check if the order of samples is the same
assert merged_rna.index.equals(merged_mir.index)

# Деление 85/15
X_BULK_TRAIN, X_BULK_TEST, Y_BULK_TRAIN, Y_BULK_TEST = train_test_split(
    merged_rna,
    merged_mir,
    test_size=0.15,
    random_state=42,
    shuffle=True
)

print(X_BULK_TRAIN.shape, X_BULK_TEST.shape)
print(Y_BULK_TRAIN.shape, Y_BULK_TEST.shape)
```

```python
import Path
here = Path(__file__).resolve().parent
workdir = here.parent
train_dir = workdir / 'ml_pipeline/data/bulk_TRAIN'

X_BULK_TRAIN.to_parquet(train_dir / 'X_BULK_TRAIN.parquet')
Y_BULK_TRAIN.to_parquet(train_dir / 'Y_BULK_TRAIN.parquet')
```

```python
test_dir = workdir / 'ml_pipeline/data/bulk_TEST'
X_BULK_TEST.to_parquet(test_dir / 'X_BULK_TEST.parquet')
Y_BULK_TEST.to_parquet(test_dir / 'Y_BULK_TEST.parquet')
```
